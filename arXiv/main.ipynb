```python
import pandas as pd
import numpy as np


with open("/Users/hayden/Documents/data471/finalProj/Final-Project-DATA471/arXiv/data/train.txt", "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

text = pd.DataFrame(lines, columns=["text"])
print(text.head())
print(text.shape)

labels = pd.read_csv("/Users/hayden/Documents/data471/finalProj/Final-Project-DATA471/arXiv/data/train.CT", header=None, names=["label"])
print(labels.head())
print(labels.shape)

text_lengths = text['text'].str.len()
print(text_lengths.head())
print(text_lengths.shape)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import CountVectorizer

X_train, X_test, y_train, y_test = train_test_split(text, labels, test_size=0.2, random_state=0)
X_train = X_train.to_numpy().ravel()
X_test = X_test.to_numpy().ravel()
y_train = y_train.to_numpy().ravel()
y_test = y_test.to_numpy().ravel()

clf = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0))
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

print(y_train.shape)

clf = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0))
clf.fit(X_train, y_train)

```

```python
from sklearn.metrics import f1_score
y_pred = clf.predict(X_test)

print(y_test.shape)
print(y_pred.shape)
print(y_pred)

score = f1_score(y_test, y_pred, average='weighted')
print(f"Binary F1 Score: {score:.3f}")
```
